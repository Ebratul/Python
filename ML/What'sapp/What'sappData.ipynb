```python
import re
import pandas as pd
```

```python
f = open("/content/drive/MyDrive/dataset/WhatsApp Chat with Niko_Ratul_Choto_Mama.txt", 'r', encoding='utf-8')
```

```python
data = f.read()
```

```python
print(data)
```

```python
data = data.replace('\u202f', ' ')
```

```python
pattern = r'\d{1,2}/\d{1,2}/\d{4},\s\d{1,2}:\d{2}\s[ap]m\s-\s'
```

```python
messages = re.split(pattern, data)[10:]
messages
```

```python
dates = re.findall(pattern, data)[9:]
dates
```

```python
df = pd.DataFrame({'UserMassage' : messages, 'message_data':dates})

df['message_data'] = pd.to_datetime(df['message_data'], format='%d/%m/%Y, %I:%M %p - ')
df.rename(columns={'message_data': 'date'}, inplace=True)
df.head()
```

```python
df.shape
```

```python
users = []
messages = []

for msg in df['UserMassage']:
    entry = re.split(r'([^:]+):\s', msg, maxsplit=1) #entry = ['', 'Apo', 'hi']

    if len(entry) > 2:
        users.append(entry[1])
        messages.append(entry[2])
    else:
        users.append('group_notification')
        messages.append(entry[0])

df['user'] = users
df['message'] = messages
df.drop(columns=['UserMassage'], inplace=True)

df.head()
```
