```python
tmdb_movie_genres = {
    28: "Action",
    12: "Adventure",
    16: "Animation",
    35: "Comedy",
    80: "Crime",
    99: "Documentary",
    18: "Drama",
    10751: "Family",
    14: "Fantasy",
    36: "History",
    27: "Horror",
    10402: "Music",
    9648: "Mystery",
    10749: "Romance",
    878: "Science Fiction",
    10770: "TV Movie",
    53: "Thriller",
    10752: "War",
    37: "Western"
}
columns = [
    'id', 'adult', 'overview', 'genres', 'title', 'release_date',
    'vote_average', 'popularity', 'vote_count', 'video',
    'original_language', 'original_title'
]

data_2d = []

for j in range(1, 501):
  response = requests.get('https://api.themoviedb.org/3/movie/popular?api_key=effa57ee63ee347383ea69b5dfbb61c3&language=en-US&page={}'.format(j)).json()
  for item in response['results']:
    genre_ids = item.get('genre_ids', [])
    genre_names = [tmdb_movie_genres.get(gid, 'Unknown') for gid in genre_ids]
    genre_string = ', '.join(genre_names)
    row = []
    for col in columns:
      if col == 'genres':
        row.append(genre_string)
      else:
        row.append(item.get(col))
      data_2d.append(row)
df = pd.DataFrame(data_2d, columns=columns)

```
