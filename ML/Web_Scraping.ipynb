```python
Processor = []
RAM = []
Display = []
Graphics = []
Feature = []
Storage = []

# unique_set = set()  # set for tracking unique keys

for item in x:
    li_list = item.find_all('li')

    p = r = d = g = f = s = 0

    for li in li_list:
        text = li.get_text(strip=True)

        if text.startswith('Processor'):
            Processor.append(text)
            p = 1
        elif text.startswith('RAM'):
            RAM.append(text)
            r = 1
        elif text.startswith('Display'):
            Display.append(text)
            d = 1
        elif text.startswith('Graphics'):
            Graphics.append(text)
            g = 1
        elif text.startswith('Feature'):
            Feature.append(text)
            f = 1
        elif text.startswith('Storage'):
            Storage.append(text)
            s = 1
        # unique_set.add(text.split()[0])

    if p == 0:
        Processor.append('None')
    if r == 0:
        RAM.append('None')
    if d == 0:
        Display.append('None')
    if g == 0:
        Graphics.append('None')
    if f == 0:
        Feature.append('None')
    if s == 0:
        Storage.append('None')
```

```python
x = soup.find_all(class_ = 'p-item')
len(x)
```

```python
x[1]
```

```python
laptop_name = []

Processor = []
RAM = []
Display = []
Graphics = []
Feature = []
Storage = []
price = []

for i in x:
    # Laptop name
    laptop_name.append(i.find_all(class_='p-item-name')[0].text.strip())

    # Short description div
    val = i.find_all('div', class_='short-description')

    # If div exists
    if val:
        li_list = val[0].find_all('li')
    else:
        li_list = []

    # Flags
    p = r = d = g = f = s = 0

    for li in li_list:
        text = li.get_text(strip=True)

        if text.startswith('Processor'):
            Processor.append(text)
            p = 1

        elif text.startswith('RAM'):
            RAM.append(text)
            r = 1

        elif text.startswith('Display'):
            Display.append(text)
            d = 1

        elif text.startswith('Graphics'):
            Graphics.append(text)
            g = 1

        elif text.startswith('Feature'):
            Feature.append(text)
            f = 1

        elif text.startswith('Storage'):
            Storage.append(text)
            s = 1

    # Missing fields fill with "None"
    if p == 0:
        Processor.append('None')
    if r == 0:
        RAM.append('None')
    if d == 0:
        Display.append('None')
    if g == 0:
        Graphics.append('None')
    if f == 0:
        Feature.append('None')
    if s == 0:
        Storage.append('None')
```
