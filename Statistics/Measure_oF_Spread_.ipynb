#Standard Deviation

```python
import pandas as pd
import seaborn as sb
from scipy.stats import norm
import matplotlib.pyplot as plt
import numpy as np
```

```python
#Generate 100 normal random integer with specified mean and std
# mean = 1000 and SD = 200
# Percentage Point Function (PPF)
draw = norm.ppf(np.random.random(100000), loc = 1000, scale = 200).astype(int)
sb.distplot(draw)
```

```python
z = []
for  i in draw:
  z.append((i-1000)/200)
```

```python
a = np.array(z)
```

```python
a.mean()
```

```python
np.round(a.mean())
```

```python
a.std()
```

```python
np.round(a.std())
```

```python
sb.distplot(a)
```

#Z score

Problem 1: Imagine a group of 300 students took a test. Rohit scored 700 out of 1000. The average score was 500 and the standard deviation was 120. Assuming that the test scores followed normal distribution find out how well Rohit performed in comparision to his peers

```python
# calculate z score

x = (700 - 500)/120
print(x)

# lookup table/ztable
# 1.6 + 0.06 = x
```

```python
# Lookup the z table
a = 0.9515
```

```python
a *100
# rohit performed better than 955 of his peers
```

```python
# n = 50/461 batsman take per sample
# m = 1000 number of time sampling done
```

```python
import numpy as np
```

```python
np.random.choice(run, size = 25)
```

```python
m  = 1000
n = 50
new = []
for i in range(m):
  x = np.random.choice(run, size = n)
  new.append(x.mean())
```

```python
len(new)
```

```python
sb.distplot(new)
```

```python
# Variance of the Sampling Distribution
run.var()/50
```
