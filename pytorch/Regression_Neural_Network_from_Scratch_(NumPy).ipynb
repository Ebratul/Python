```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df_train = pd.read_csv('/content/random-linear-regression/train.csv')
df_test = pd.read_csv('/content/random-linear-regression/test.csv')
```

```python
df_train.sample(4)
```

```python
df_train = df_train.fillna(df_train.mean(numeric_only=True))
# df_train['y'] = df_train['y'].fillna(df_train['y'].mean())
```

```python

X_trian = df_train.drop('y', axis=1)
y_train = df_train['y']
X_test = df_test.drop('y', axis=1)
y_test = df_test['y']
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
# Scaling
scaler = StandardScaler()
X_train = scaler.fit_transform(X_trian)
X_test = scaler.transform(X_test)
```

```python
import numpy as np
```

```python
np.random.seed(42)

n_samples, n_features = X_train.shape
hidden_units = 10
lr = 0.01
epochs = 2000
```

```python
# Weights & biases
W1 = np.random.randn(n_features, hidden_units) * 0.01
b1 = np.zeros((1, hidden_units))

W2 = np.random.randn(hidden_units, 1) * 0.01
b2 = np.zeros((1, 1))
```

```python
print(X_train.shape)
print(W1.shape)
print(W2.shape)
print(y_train.shape)
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def sigmoid_derivative(z):
    s = sigmoid(z)
    return s * (1 - s)
```

```python
y_train = y_train.to_numpy().reshape(-1, 1)
y_test  = y_test.to_numpy().reshape(-1, 1)
```

```python
# Training loop
for epoch in range(epochs):

    # ---- Forward pass ----
    z1 = X_train.dot(W1) + b1
    A1 = sigmoid(z1)

    z2 = A1.dot(W2) + b2
    y_pred = z2

    # ---- Loss (MSE) ----
    loss = np.mean((y_pred - y_train) ** 2)

    # ---- Backpropagation ----
    dz2 = 2 * (y_pred - y_train) / n_samples
    dW2 = A1.T.dot(dz2)
    db2 = np.sum(dz2, axis=0, keepdims=True)

    dA1 = dz2.dot(W2.T)
    dz1 = dA1 * sigmoid_derivative(z1)

    dW1 = X_train.T.dot(dz1)
    db1 = np.sum(dz1, axis=0, keepdims=True)

    # ---- Gradient descent ----
    W2 -= lr * dW2
    b2 -= lr * db2
    W1 -= lr * dW1
    b1 -= lr * db1

    if epoch % 100 == 0:
        print(f"Epoch {epoch}, Loss: {loss}")
```

```python
# Evaluation
Z1_test = X_test @ W1 + b1
A1_test = sigmoid(Z1_test)
y_test_pred = A1_test.dot(W2) + b2

test_mse = np.mean((y_test_pred - y_test) ** 2)
print("Test MSE:", test_mse)
```
